```python
def normalize_url(url):
    url = url.strip()
    
    if not url.startswith("http://") and not url.startswith("https://"):
        url = "https://" + url
    
    if validators.url(url):
        return url
    else:
        return None

urls_input = input("Unesite web sajtove odvojene zarezom:\n")
raw_urls = [url.strip() for url in urls_input.split(",") if url.strip() != ""]

urls = []
invalid_urls = []

for url in raw_urls:
    normalized = normalize_url(url)
    if normalized:
        urls.append(normalized)
    else:
        invalid_urls.append(url)

print(f"\nValidni URL-ovi:\n{urls}")

if invalid_urls:
    print(f"\nNevalidni URL-ovi (preskočeni):\n{invalid_urls}")
```

```python
def check_website(url, timeout=5):
    result = {"URL": url, "Status Code": None, "Response Time (s)": None, "Status": None}
    try:
        start = time.time()
        response = requests.get(url, timeout=timeout)
        end = time.time()
        
        result["Status Code"] = response.status_code
        result["Response Time (s)"] = round(end - start, 3)
        
        if response.status_code == 200:
            result["Status"] = "UP"
        elif response.status_code == 404:
            result["Status"] = "NOT FOUND"
        elif response.status_code >= 500:
            result["Status"] = "SERVER ERROR"
        else:
            result["Status"] = "UNKNOWN"
    except requests.exceptions.Timeout:
        result["Status"] = "DOWN (Timeout)"
    except requests.exceptions.RequestException:
        result["Status"] = "DOWN (Error)"
    
    return result

results = []

for url in urls:
    res = check_website(url)
    results.append(res)
    print(f"{res['URL']}: {res['Status']} ({res['Status Code']}) - {res['Response Time (s)']}s")

df_results = pd.DataFrame(results)
df_results
```

```python
def calculate_score(row):
    # Ako nema response time → loš score
    if pd.isna(row['Response Time (s)']) or row['Response Time (s)'] == 0:
        return 0
    
    base_score = (1 / row['Response Time (s)']) * 100
    
    # Penalizacija
    if row['Status'] != "UP":
        base_score *= 0.3  # veliki penal za greške
    
    return round(base_score, 2)

# Dodavanje score kolone
df_results['Performance Score'] = df_results.apply(calculate_score, axis=1)

# Ranking (veći score = bolji)
df_results = df_results.sort_values(by='Performance Score', ascending=False).reset_index(drop=True)

print("\n🏆 PERFORMANCE RANKING:")
display(df_results[['URL', 'Status', 'Response Time (s)', 'Performance Score']])
```

```python
HISTORY_FILE = "history.csv"

def save_history(df):
    df_copy = df.copy()
    df_copy["Timestamp"] = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    
    if os.path.exists(HISTORY_FILE):
        df_copy.to_csv(HISTORY_FILE, mode='a', header=False, index=False)
    else:
        df_copy.to_csv(HISTORY_FILE, index=False)

save_history(df_results)
print("📁 Istorija sačuvana")

def load_history():
    if os.path.exists(HISTORY_FILE):
        try:
            # Ignoriši linije koje ne odgovaraju formatu
            df_hist = pd.read_csv(HISTORY_FILE, on_bad_lines='skip')
            # Ako postoji kolona Timestamp, konvertuj u datetime
            if "Timestamp" in df_hist.columns:
                df_hist["Timestamp"] = pd.to_datetime(df_hist["Timestamp"], errors='coerce')
            return df_hist
        except Exception as e:
            print(f"Greška pri učitavanju istorije: {e}")
            return pd.DataFrame()
    else:
        print("Nema istorije još.")
        return pd.DataFrame()

df_history = load_history()
df_history.head()
```

```python
os.makedirs("charts", exist_ok=True)

# Bar chart
plt.figure(figsize=(10,6))
plt.bar(df_results['URL'], df_results['Response Time (s)'].fillna(0))
plt.xticks(rotation=45, ha='right')
plt.ylabel('Response Time (s)')
plt.title('Vreme odziva po sajtu')
plt.tight_layout()
bar_path = "charts/bar_chart.png"
plt.savefig(bar_path)
plt.show()

# Pie chart
status_counts = df_results['Status'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(status_counts, labels=status_counts.index, autopct='%1.1f%%')
plt.title('Distribucija statusa sajtova')
pie_path = "charts/pie_chart.png"
plt.savefig(pie_path)
plt.show()

# Histogram
plt.figure(figsize=(10,6))
status_codes = df_results['Status Code'].fillna(0)
plt.hist(status_codes, bins=range(0, 600, 50))
plt.xlabel('HTTP Status Code')
plt.ylabel('Broj sajtova')
plt.title('Frekvencija HTTP status kodova')
hist_path = "charts/hist_chart.png"
plt.savefig(hist_path)
plt.show()

def plot_trend(df_history):
    if df_history.empty:
        print("Nema podataka za trend.")
        return None
    
    plt.figure(figsize=(12,6))
    
    for url in df_history['URL'].unique():
        df_url = df_history[df_history['URL'] == url].sort_values("Timestamp")
        plt.plot(df_url["Timestamp"], df_url["Response Time (s)"], marker='o', label=url)
    
    plt.xlabel("Vreme")
    plt.ylabel("Response Time (s)")
    plt.title("Trend brzine kroz vreme")
    plt.legend()
    plt.xticks(rotation=45)
    plt.tight_layout()
    
    trend_path = "charts/trend_chart.png"
    plt.savefig(trend_path)
    plt.show()
    
    return trend_path

trend_chart_path = plot_trend(df_history)

def calculate_trend(df_history):
    trends = []

    for url in df_history['URL'].unique():
        df_url = df_history[df_history['URL'] == url].sort_values("Timestamp")
        
        if len(df_url) < 2:
            trend = "Nedovoljno podataka"
        else:
            first = df_url.iloc[0]['Response Time (s)']
            last = df_url.iloc[-1]['Response Time (s)']
            
            if last < first:
                trend = "🚀 Poboljšanje"
            elif last > first:
                trend = "🐢 Pogoršanje"
            else:
                trend = "➖ Nema promene"
        
        trends.append({"URL": url, "Trend": trend})
    
    return pd.DataFrame(trends)

df_trends = calculate_trend(df_history)

print("\n📈 TREND ANALIZA:")
display(df_trends)
```

```python
def advanced_health_score(url, df_history):
    # Uptime %
    df_url = df_history[df_history['URL'] == url]
    if df_url.empty:
        uptime = 0
    else:
        uptime = (df_url['Status'] == 'UP').mean() * 100
    
    # Stabilnost (std dev response time)
    stability = 100 if df_url.empty else max(0, 100 - np.std(df_url['Response Time (s)'].fillna(0))*10)
    
    # Status codes kroz vreme
    code_penalty = 100
    if not df_url.empty:
        errors = df_url['Status Code'].fillna(0)
        code_penalty = 100 - (errors[errors>=400].count() / len(errors) * 50)
    
    # Kombinovani score
    score = round((uptime*0.4 + stability*0.3 + code_penalty*0.3),2)
    
    if score >= 90:
        label = "Excellent 🚀"
    elif score >= 70:
        label = "Good 👍"
    elif score >= 50:
        label = "Needs Optimization ⚠️"
    else:
        label = "Poor ❌"
    
    return pd.Series({"Health Score": score, "Health Label": label})

# SEO & basic analiza
def seo_basic_analysis(url):
    try:
        resp = requests.get(url, timeout=5)
        content = resp.text
        soup = BeautifulSoup(content, "html.parser")
        
        title = bool(soup.title and soup.title.string.strip())
        meta_desc = bool(soup.find("meta", attrs={"name":"description"}))
        page_size = len(resp.content)/1024  # KB
        num_images = len(soup.find_all("img"))
        num_scripts = len(soup.find_all("script"))
        
        return pd.Series({
            "Has Title": title,
            "Has Meta Description": meta_desc,
            "Page Size (KB)": round(page_size,2),
            "Images": num_images,
            "Scripts": num_scripts
        })
    except:
        return pd.Series({
            "Has Title": False,
            "Has Meta Description": False,
            "Page Size (KB)": None,
            "Images": None,
            "Scripts": None
        })

# Dodavanje u df_results
df_results[['Health Score', 'Health Label']] = df_results['URL'].apply(lambda u: advanced_health_score(u, df_history))
seo_df = df_results['URL'].apply(seo_basic_analysis)
df_results = pd.concat([df_results, seo_df], axis=1)

print("\n🔍 Health & SEO Analysis:")
display(df_results[['URL','Health Score','Health Label','Has Title','Has Meta Description','Page Size (KB)','Images','Scripts']])
```

```python
# Napravi kopiju DataFrame-a i ukloni sve emotikone
df_pdf = df_results.copy()
df_pdf['Health Label PDF'] = df_pdf['Health Label'].str.replace(r'[^\x00-\x7F]+', '', regex=True)

# Isto za trend kolonu
df_trends_pdf = df_trends.copy()
df_trends_pdf['Trend PDF'] = df_trends_pdf['Trend'].str.replace(r'[^\x00-\x7F]+', '', regex=True)

from fpdf import FPDF

def generate_pdf(df, filename="website_report.pdf"):
    pdf = FPDF()
    pdf.set_auto_page_break(auto=True, margin=15)
    pdf.add_page()

    pdf.set_font("Arial", "B", 16)
    pdf.cell(0, 10, "Website Status Report", ln=True, align='C')

    pdf.set_font("Arial", "", 12)
    pdf.cell(0, 10, f"Date & Time: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}", ln=True, align='C')
    pdf.ln(10)

    pdf.set_font("Arial", "B", 10)
    col_width = pdf.w / 6
    headers = ["URL", "Status Code", "Response Time", "Status", "Score", "Health", "Label"]
    for h in headers:
        pdf.cell(col_width, 10, h, 1)
    pdf.ln()

    pdf.set_font("Arial", "", 9)
    for _, row in df.iterrows():
        pdf.cell(col_width, 10, str(row['URL'])[:25], 1)
        pdf.cell(col_width, 10, str(row['Status Code']), 1)
        pdf.cell(col_width, 10, str(row['Response Time (s)']), 1)
        pdf.cell(col_width, 10, str(row['Status']), 1)
        pdf.cell(col_width, 10, str(row['Performance Score']), 1)
        pdf.cell(col_width, 10, str(row['Health Score']), 1)
        pdf.cell(col_width, 10, str(row['Health Label PDF']), 1)  # Unicode-free
        pdf.ln()

    # Charts
    pdf.add_page()
    pdf.set_font("Arial", "B", 14)
    pdf.cell(0, 10, "Charts:", ln=True)
    pdf.image(bar_path, w=180)
    pdf.ln(5)
    pdf.image(pie_path, w=150)
    pdf.ln(5)
    pdf.image(hist_path, w=180)

    # Trend chart
    if trend_chart_path:
        pdf.add_page()
        pdf.cell(0, 10, "Trend kroz vreme:", ln=True)
        pdf.image(trend_chart_path, w=180)

    # Trend tekst – Unicode uklonjen
    pdf.ln(10)
    pdf.set_font("Arial", "", 10)
    for _, row in df_trends_pdf.iterrows():
        pdf.cell(0, 10, f"{row['URL']} -> {row['Trend PDF']}", ln=True)

    pdf.output(filename)
    print(f"\nPDF '{filename}' kreiran!")
```

```python
choice = input("Da li želite da generišete PDF izveštaj? (yes/no): ").strip().lower()

if choice == "yes":
    generate_pdf(df_pdf, filename="website_report.pdf")
```
