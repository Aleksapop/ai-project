```python
class Mnozenje:
    def mnozenje(self, test_dva, broj):
        prvi = test_dva[0][0]
        drugi = test_dva[1][1]
        rezultat = broj * prvi + broj * drugi
        return(rezultat)
```

```python
mno = Mnozenje()

broj = int(input("Unesi broj lepoto: "))

rez = mno.mnozenje(test_dva, broj)

print(rez)
```

```python
# napisi klasu koja sabira dva broja
class Matematike:
    def __init__(self, a, b):
        self.a = a
        self.b = b
        
    def sabira(self, a = None, b = None):
        if a is not None:
            self.a = a
        if b is not None:
            self.b = b
        rezultat = self.a + self.b
        return rezultat
```

```python
m = Matematike(10, 20)
print(m.sabira())
m.sabira(10, 10)
```
