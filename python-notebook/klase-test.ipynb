```python
# funckiju koja mnozi tri broja

def pomnozi(broj, broj_dva, broj_tri):
    rezultat = broj * broj_dva * broj_tri
    return(rezultat)
```

```python
rez = pomnozi(2, 2, 2)
print(rez)
```
