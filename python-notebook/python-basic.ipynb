```python
reci = ["mačka", "pas", "auto", "stolica", "sunce", "oko", "rad"]
def dodaj(reci):
    manje = []
    tacno = []
    vise = []

    for slovo in reci:
        if len(slovo) < 4:
            manje.append(slovo)
        elif len(slovo) == 4:
            tacno.append(slovo)
        else:
            vise.append(slovo)
    print(manje)
    print(tacno)
    print(vise)
    print(len(manje))
    print(len(tacno))
    print(len(vise))
dodaj(reci)
```

```python
def broj_slova(reci):
    parne_duzine = []
    neparne_duzine =[]
    pocinje_sa_a = []

    for broj in reci:
        if len(broj) % 2 == 0:
            parne_duzine.append(broj)
        else:
            neparne_duzine.append(broj)
        if broj.startswith("a") or broj.startswith("A"):
            pocinje_sa_a.append(broj)
            
    print("Paran broj slova:", parne_duzine)
    print("Neparan broj slova:", neparne_duzine)
    print("Slova koja pocinju sa a ili A:", pocinje_sa_a)
    print("reci sa parnim brojem slova ima:", len(parne_duzine))
    print("reci sa neparnim brojem slova ima:", len(neparne_duzine))
    print("imena koja pocinju sa a ili sa A:", len(pocinje_sa_a))
```

```python
test = ["auto", "stolica", "ana", "pas", "Aleksandar", "oko"]
broj_slova(test)
```

```python
def dodavanje(test):
    poz_parni = []
    poz_neparni = []
    negativni = []
    nula = []

    for item in test:
        if item % 2 == 0 and item > 0:
            poz_parni.append(item)
        elif item % 2 and item > 0:
            poz_neparni.append(item)
        elif item == 0:
            nula.append(item)
        if item < 0:
            negativni.append(item)
    print("Pozitivni i parni brojevi:", poz_parni, "i ima ih", len(poz_parni))
    print("Pozitivni i neparni brojevi:", poz_neparni, "i ima ih", len(poz_neparni))
    print("Neparni brojevi su:", negativni, "i ima ih", len(negativni))
    print("Nule:", nula, "i ima ih", len(nula))
```

```python
brojevi = []
for lista in range(-10, 31):
    brojevi.append(lista)

dodavanje(brojevi)
```

```python
def vezba(ime):
    kratke = []
    srednje = []
    duge = []
    pocinje_sa_s = []

    for item in ime:
        if len(item) < 5:
            kratke.append(item)
        elif len(item) == 5:
            srednje.append(item)
        else:
            duge.append(item)
        if item.startswith("s") or item.startswith("S"):
            pocinje_sa_s.append(item)
            
    print("reci sa manje od 5 slova:", kratke, "i ima ih", len(kratke))
    print("reci koje imaju 5 slova:", srednje, "i ima ih", len(srednje))
    print("reci koje imaju vise od 5 slova:", duge, "i ima ih", len(duge))
    print("reci koje pocinju malim ili velikim slovom s:", pocinje_sa_s, "i ima ih", len(pocinje_sa_s))
```

```python
reci = ["sunce", "pas", "stolica", "auto", "Sneg", "krava"]
vezba(reci)
```

```python
def reci(rec):
    duge_a = []
    kratke_m = []
    ostale = []

    for item in rec:
        if len(item) > 4 and (item.startswith("a") or item.startswith("A")):
            duge_a.append(item)
        elif len(item) <= 4 and (item.startswith("m") or item.startswith("M")):
            kratke_m.append(item)
        else:
            ostale.append(item)

    print("Vise od 4 slova i pocinje sa a", duge_a, "i ima ih", len(duge_a))
    print("manje od 4 slova i pocinje sa a", kratke_m, "i ima ih", len(kratke_m))
    print("ostale reci", ostale, "i ima ih", len(ostale))
```
