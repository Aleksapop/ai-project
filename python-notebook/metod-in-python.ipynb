```python
def izvlaci(o):
    lista = []

    for item in o:
        lista.append(o[item])
    return lista

def dodaj(o, k, v):
    o[k] = v
```

```python
objekat = {"ime": "aleksa", "prezime": "popovic"} 
izvlaci(objekat)
dodaj(objekat, "adresa", "selo")
print(objekat)
```

```python
#napravi funkciju koja ce da uzme neku listu i iz te liste ce da izbaci parametre koji se nalaze u vrednosti objekta

def izbaci_o(lista):
    lista_new = []

    for item in lista:
        if not isinstance(item, dict):
            lista_new.append(item)
    return lista_new
```

```python
lista = [1, 2, {"a": 3, "b": 4}, 5, 6]
izbaci_o(lista)
```

```python
def izbaci(lista):
    lista_new = []

    for item in lista:
        if isinstance(item, int):
            lista_new.append(item)
        elif isinstance(item, str):
            lista_new.append(item)
    return lista_new 
```

```python
lista = [1, 2, {"a": 3, "b": 4}, 5, 6]
izbaci(lista)
```

```python
def izbaci(lista):
    lista_new = []

    for item in lista:
        if isinstance(item, dict):
            for vrednost in item.values():
                lista_new.append(vrednost)
        else:
            lista_new.append(item)
    return lista_new
```

```python
lista = [1, 2, {"a": 3, "b": 4}, 5, 6]
izbaci(lista)
```

```python
def izbaci(lista, objekat):
    vrednosti = objekat.values()
    lista_new = []

    for item in lista:
        if item not in vrednosti:
            lista_new.append(item)
    return lista_new
```

```python
lista = [1, 2, 3, 4, 5]
objekat = {
    "a": 2,
    "b": 4
}
izbaci(lista, objekat)
```
