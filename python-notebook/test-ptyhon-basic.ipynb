```python
#2 ptyhon napravi program koji ce da printa brojeve od 1 do 50 samo parne brojeve sa 3 5 i 9 koji su sa 3 i 5 treba da kaze deljivi su sa oba broja

for item in range(1, 51):
    if item % 3 == 0 and item % 9 == 0:
        print(item, "broj je deljiv sa 9 i sa 3")
    elif item % 9 == 0:
        print(item, "broj je deljiv sa 9")
    elif item % 3 == 0 and item % 5 == 0:
        print(item, "broj je deljiv sa 3 i sa 5")
    elif item % 3 == 0:
        print(item, "broj je deljiv sa 3")
    elif item % 5 == 0:
        print(item, "broj je deljiv sa 5")
```

```python
def broj_samoglasnike(recenica):
    samoglasnici = ["a", "e", "i", "o", "u"]
    zbir = 0
    
    for i in samoglasnici:
        zbir = zbir + recenica.count(i)
    return zbir 
```

```python
def solution(number):
    if number < 0:
       return 0
    zbir = 0
    for i in range(1, number):
        if i % 3 == 0 and i % 5 == 0:
            zbir = zbir + i
        elif i % 3 == 0:
            zbir = zbir + i
        elif i % 5 == 0:
            zbir = zbir + i
     
    return zbir
```

```python
solution(10)
```
