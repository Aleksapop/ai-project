```python
#napisi mi funkciju koja ce da prima objekat i listu kao parametre i da kombanuje listu i objekat u novu listu 
#koja ce da sadrzi sve, ako su neki ajtemi duplicirani da ne ulaze u novu listu

def prima(objekat, lista):
    lista_new = []

    for item in objekat.values():
        lista_new.append(item)

    for i in lista:
        lista_new.append(i)
        
    return list(set(lista_new))
        
reci = [1, 2, 3, "aleksa", "aleksa", 1]
osoba = {"ime": "aleksa", "prezime": "popovic"}

prima(osoba, reci)
```

```python
def rgb(r, g, b):
    rgb = [r, g, b]

    broj = 0
    rgb_keys = {"A": 10, "B": 11, "C": 12, "D": 13, "E": 14, "F": 15}
    heksa = ""
    for item in rgb:
        if item > 255:
            item = 255
        if item < 0:
            item = 0
        rez = int(item / 16)
        ostatak = item % 16

        
        if rez < 10:
            heksa = heksa + str(rez)
        for i in rgb_keys:
            if rgb_keys[i] == rez:
                heksa = heksa + i


        if ostatak < 10:
            heksa = heksa + str(ostatak)
        for i in rgb_keys: 
            if rgb_keys[i] == ostatak:
                heksa = heksa + i
                

    return heksa
```

```python
a = "abc"
lista = []
lista_rezultat = []

for item in a:
    lista.append(item)

for i in range(len(lista)):
    prvi = lista[i]
    for j in range(len(lista)):
        if j != i: 
            drugi = lista[j]
            for k in range(len(lista)):
                if k != i and k != j: 
                    treci = lista[k]
                    kombinacija = prvi + drugi + treci
                    if kombinacija not in lista_rezultat:
                        lista_rezultat.append(kombinacija)

print(lista_rezultat)
```

```python
a = "abc"
lista = list(a)
rezultat = [[]]  

for c in lista:
    novi_rezultat = []
    for perm in rezultat:
        for i in range(len(perm) + 1):
            novi_perm = perm[:i] + [c] + perm[i:]
            novi_rezultat.append(novi_perm)
    rezultat = novi_rezultat

lista_rezultat = []
for perm in rezultat:
    perm_str = "".join(perm)
    if perm_str not in lista_rezultat:  
        lista_rezultat.append(perm_str)

print(lista_rezultat)
```
